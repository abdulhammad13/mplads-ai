```python
def clean_columns(df):
    df = df.copy()
    df.columns = (
        df.columns
        .str.strip()
        .str.lower()
        .str.replace(" ", "_")
        .str.replace("'", "")
        .str.replace("(", "")
        .str.replace(")", "")
        .str.replace("₹", "rs", regex=False)
    )
    return df
```

## Convert Date Columns

```python
def convert_dates(df, columns):
    df = df.copy()
    
    for column in columns:
        df[column] = pd.to_datetime(df[column], errors="coerce")
    
    return df
```

```python
calamity = convert_dates(calamity, ["date_of_consent"])

expenditure = convert_dates(expenditure, ["expenditure_date"])

completed = convert_dates(completed, ["completion_date"])

recommended = convert_dates(
    recommended,
    ["recommended_date", "sanction_date"]
)

sanctioned = convert_dates(
    sanctioned,
    ["recommended_date", "sanction_date"]
)
```

## Clean Monetary Columns

```python
def clean_money(series):
    return (
        series.astype(str)
        .str.replace(",", "", regex=False)
        .str.replace("₹", "", regex=False)
        .str.strip()
        .replace({"": pd.NA, "nan": pd.NA})
        .pipe(pd.to_numeric, errors="coerce")
    )
```

## Text Standardization

```python
text_columns = [
    "state",
    "ida",
    "honble_members_of_parliament",
    "constituency",
    "work",
    "work_description",
    "vendor_name",
    "work_category",
    "work_status",
    "payment_status"
]

def clean_text(series):
    return (
        series.astype("string")
        .str.strip()
        .str.replace(r"\s+", " ", regex=True)
    )
```

```python
for df in [allocations, calamity, expenditure, completed, recommended, sanctioned]:
    for column in text_columns:
        if column in df.columns:
            df[column] = clean_text(df[column])
```

```python
# 1a. Drop trailing "Grand Total" rows — they have blank identifier fields
def drop_summary_rows(df, id_cols):
    mask = df[id_cols].isna().all(axis=1)
    if mask.any():
        print(f"Dropped {mask.sum()} summary row(s)")
    return df[~mask].reset_index(drop=True)

allocations = drop_summary_rows(allocations, ["state", "honble_members_of_parliaments"])
calamity    = drop_summary_rows(calamity, ["calamity_type", "honble_members_of_parliament"])
expenditure = drop_summary_rows(expenditure, ["state", "ida"])
completed   = drop_summary_rows(completed, ["state", "ida"])
recommended = drop_summary_rows(recommended, ["state", "ida"])
sanctioned  = drop_summary_rows(sanctioned, ["state", "ida"])
```

```python
# 1b. Fix case/whitespace duplicates (e.g. "Completed" vs "completed") — collapse to most common form
def canonicalize(series):
    s = series.astype("string").str.strip().str.replace(r"\s+", " ", regex=True)
    return s.groupby(s.str.lower()).transform(lambda x: x.mode().iloc[0])

cat_cols = ["state", "ida", "honble_members_of_parliament", "constituency",
            "work_category", "work_status", "payment_status", "vendor_name"]

for df in [allocations, calamity, expenditure, completed, recommended, sanctioned]:
    for col in cat_cols:
        if col in df.columns:
            df[col] = canonicalize(df[col])
```

```python
def make_key(df, cols):
    """Build a normalized composite string key from a list of columns. Robust to NaN/mixed dtypes."""
    key = df[cols[0]].fillna("").astype(str).str.strip()
    for col in cols[1:]:
        key = key + "|" + df[col].fillna("").astype(str).str.strip()
    return key

def check_key_uniqueness(df, key_col, name):
    n = len(df)
    u = df[key_col].nunique()
    print(f"{name}: {n} rows, {u} unique keys, {n - u} duplicate-key rows")
```

```python
link_cols = [
    "state", "ida", "honble_members_of_parliament",
    "constituency", "work_category", "work_description", "recommended_date"
]

recommended["link_key"] = make_key(recommended, link_cols)
sanctioned["link_key"]  = make_key(sanctioned, link_cols)

check_key_uniqueness(recommended, "link_key", "recommended")
check_key_uniqueness(sanctioned, "link_key", "sanctioned")

sanc_subset = sanctioned[["link_key", "sanction_amount__rs_", "sanction_date", "work_status"]].rename(
    columns={"sanction_date": "sanction_date_actual"}
)

rec_sanc = recommended.merge(sanc_subset, on="link_key", how="outer", indicator=True)

print("\nMatch breakdown:")
print(rec_sanc["_merge"].value_counts())
```

```python
complete_link_cols = [
    "state", "ida", "honble_members_of_parliament",
    "constituency", "work_category", "work_description"
]

rec_sanc["link_key_2"] = make_key(rec_sanc, complete_link_cols)
completed["link_key_2"] = make_key(completed, complete_link_cols)

check_key_uniqueness(completed, "link_key_2", "completed")

master = rec_sanc.merge(
    completed[["link_key_2", "completion_date", "amount_disbursed__rs_"]],
    on="link_key_2", how="left", indicator="completion_match"
)

print("\nCompletion match breakdown:")
print(master["completion_match"].value_counts())
# both      = work has a completion record
# left_only = still recommended/sanctioned but not marked complete
```

```python
# Expenditure has no work_category/work_description — it only has 'work', so key on that instead
exp_link_cols = ["state", "ida", "honble_members_of_parliament", "constituency", "work"]

expenditure["link_key_3"] = make_key(expenditure, exp_link_cols)
master["link_key_3"]      = make_key(master, exp_link_cols)   # master still has 'work' from recommended

exp_agg = expenditure.groupby("link_key_3").agg(
    total_disbursed=("fund_disbursed_amount__rs_", "sum"),
    num_transactions=("fund_disbursed_amount__rs_", "count"),
    num_vendors=("vendor_name", "nunique"),
    first_payment=("expenditure_date", "min"),
    last_payment=("expenditure_date", "max"),
).reset_index()

master = master.merge(exp_agg, on="link_key_3", how="left")

print("Works with at least one expenditure record:",
      master["total_disbursed"].notna().sum(), "/", len(master))
```

```python
alloc_key_cols = ["state", "honble_members_of_parliaments", "constituency"]
allocations["mp_key"] = make_key(allocations, alloc_key_cols)

master["mp_key"] = make_key(
    master.rename(columns={"honble_members_of_parliament": "honble_members_of_parliaments"}),
    alloc_key_cols
)

master = master.merge(
    allocations[["mp_key", "allocated_amount__rs_"]],
    on="mp_key", how="left"
)

print("Works with a matched MP allocation record:",
      master["allocated_amount__rs_"].notna().sum(), "/", len(master))
```

```python
# ============================================================
# MASTER DATASET — DERIVED FEATURES
# ============================================================

# 1. Financial ratios
master["disbursement_to_sanction_ratio"] = (
    master["total_disbursed"] / master["sanction_amount__rs_"]
).replace([float("inf"), -float("inf")], pd.NA)

master["utilization_pct"] = (
    master["total_disbursed"] / master["allocated_amount__rs_"] * 100
).replace([float("inf"), -float("inf")], pd.NA)


# 2. Time-based features
master["days_rec_to_sanction"] = (
    master["sanction_date"] - master["recommended_date"]
).dt.days

master["days_sanction_to_complete"] = (
    master["completion_date"] - master["sanction_date"]
).dt.days


# 3. Vendor concentration
vc = (
    expenditure
    .groupby(
        ["honble_members_of_parliament", "vendor_name"],
        dropna=False
    )
    .size()
    .reset_index(name="transaction_count")
)

vendor_share = (
    vc.groupby("honble_members_of_parliament")["transaction_count"]
      .apply(lambda x: x.max() / x.sum() if x.sum() > 0 else pd.NA)
)

master["top_vendor_share"] = (
    master["honble_members_of_parliament"]
    .map(vendor_share)
)


# 4. Potential near-duplicate works
#    Use a safe key and ignore rows where the key cannot be formed.
master["desc_key"] = (
    master["ida"].fillna("").astype("string").str.strip()
    + "|"
    + master["work_description"].fillna("").astype("string").str.strip()
)

valid_desc_key = (
    master["ida"].notna()
    & master["work_description"].notna()
    & master["recommended_date"].notna()
)

duplicate_counts = master.groupby("desc_key")["desc_key"].transform("size")

date_window = (
    master.groupby("desc_key")["recommended_date"]
    .transform(lambda x: (x.max() - x.min()).days)
)

master["is_near_duplicate"] = (
    valid_desc_key
    & (duplicate_counts > 1)
    & (date_window <= 30)
)


# 5. Basic data-quality flags
master["invalid_rec_sanction_sequence"] = (
    master["days_rec_to_sanction"] < 0
)

master["invalid_sanction_completion_sequence"] = (
    master["days_sanction_to_complete"] < 0
)

master["disbursement_exceeds_sanction"] = (
    master["disbursement_to_sanction_ratio"] > 1
)


# 6. Remove the old misleading feature name if it exists
if "cost_overrun_ratio" in master.columns:
    master.drop(columns=["cost_overrun_ratio"], inplace=True)


# 7. Compact feature check
feature_columns = [
    "total_disbursed",
    "num_transactions",
    "num_vendors",
    "first_payment",
    "last_payment",
    "allocated_amount__rs_",
    "sanction_amount__rs_",
    "disbursement_to_sanction_ratio",
    "utilization_pct",
    "days_rec_to_sanction",
    "days_sanction_to_complete",
    "top_vendor_share",
    "is_near_duplicate",
    "invalid_rec_sanction_sequence",
    "invalid_sanction_completion_sequence",
    "disbursement_exceeds_sanction"
]

master[feature_columns].head()
```

```python
# ============================================================
# RISK FLAGS + RISK SCORE
# ============================================================

# 1. Financial anomaly:
#    Disbursed amount is more than 105% of sanctioned amount.
master["flag_overrun"] = (
    master["disbursement_to_sanction_ratio"] > 1.05
)


# 2. Date anomaly:
#    Completion appears before sanction.
master["flag_bad_dates"] = (
    master["completion_date"] < master["sanction_date"]
)


# 3. Allocation/utilization anomaly:
#    Here we use the existing utilization_pct correctly as a percentage.
#    >100 means total disbursed exceeds the allocated amount.
master["flag_over_allocation"] = (
    master["utilization_pct"] > 100
)


# 4. Potential duplicate/overlap flag
master["flag_duplicate"] = (
    master["is_near_duplicate"].fillna(False)
)


# 5. Cost outlier within each work category
def iqr_flag(series):
    q1 = series.quantile(0.25)
    q3 = series.quantile(0.75)
    iqr = q3 - q1

    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr

    return (series < lower) | (series > upper)


master["flag_cost_outlier"] = (
    master
    .groupby("work_category")["sanction_amount__rs_"]
    .transform(iqr_flag)
)


# 6. Duration outlier within each work category
master["flag_duration_outlier"] = (
    master
    .groupby("work_category")["days_sanction_to_complete"]
    .transform(iqr_flag)
)


# ------------------------------------------------------------
# 7. Combine all flags
# ------------------------------------------------------------

flags = [
    "flag_overrun",
    "flag_bad_dates",
    "flag_over_allocation",
    "flag_duplicate",
    "flag_cost_outlier",
    "flag_duration_outlier"
]


# Convert True/False to 1/0 and calculate 0–100 risk score
master["risk_score"] = (
    master[flags]
    .fillna(False)
    .astype(int)
    .sum(axis=1)
    / len(flags)
    * 100
)


# ------------------------------------------------------------
# 8. Risk category
# ------------------------------------------------------------

master["risk_level"] = pd.cut(
    master["risk_score"],
    bins=[-1, 20, 50, 70, 100],
    labels=["Low", "Medium", "High", "Critical"]
)


# ------------------------------------------------------------
# 9. Show highest-risk records
# ------------------------------------------------------------

result_columns = [
    "state",
    "ida",
    "honble_members_of_parliament",
    "work_description",
    "risk_score",
    "risk_level"
] + flags

master.sort_values(
    "risk_score",
    ascending=False
)[result_columns].head(20)
```
